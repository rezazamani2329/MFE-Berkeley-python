## 2. Imports

```python
import joblib
import numpy as np
from sklearn.ensemble import RandomForestRegressor
```

## 3. Generate Synthetic Training Data
`RandomState(42)` seeds the RNG for reproducibility — running this notebook multiple times will always produce the same data.

```python
# X: 100 samples × 4 features (values in range 0–1)
X = np.random.RandomState(42).rand(100, 4)

# y: 100 continuous target values (range 0–1)
y = np.random.RandomState(42).rand(100)

print(f'X shape : {X.shape}  (samples × features)')
print(f'y shape : {y.shape}  (target values)')
print(f'\nFirst 5 rows of X:\n{X[:5]}')
print(f'\nFirst 5 values of y:\n{y[:5]}')
```

## 4. Train the Random Forest Regressor
Builds `n_estimators` decision trees and averages their predictions. `random_state=42` makes the model reproducible.

```python
model = RandomForestRegressor(n_estimators=10, random_state=42)
model.fit(X, y)

print(f'Training complete.')
print(f'  n_estimators    : {model.n_estimators}')
print(f'  n_features_in_  : {model.n_features_in_}')
print(f'  Training score  : {model.score(X, y):.4f}')
```

## 5. Inspect Feature Importances

```python
for i, importance in enumerate(model.feature_importances_):
    print(f'  Feature {i}: {importance:.4f}')
```

## 6. Save the Model to Disk

```python
joblib.dump(model, 'model.joblib')
print(f'✓ Model saved to model.joblib  |  n_features_in_: {model.n_features_in_}')
```

## 7. Verify — Reload & Run a Sample Prediction

```python
loaded_model = joblib.load('model.joblib')

sample = np.array([[0.5, 0.3, 0.8, 0.1]])
prediction = loaded_model.predict(sample)[0]

print(f'Sample input  : {sample[0].tolist()}')
print(f'Prediction    : {prediction:.4f}')
print(f'Model reloaded successfully ✓')
```
